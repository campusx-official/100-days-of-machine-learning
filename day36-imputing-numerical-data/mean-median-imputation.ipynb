## Univariate vs Multivariate Imputation

| Feature | Univariate Imputer | Multivariate Imputer |
|---------|--------------------|----------------------|
| **Definition** | Imputes each feature independently. | Imputes a feature using information from other features. |
| **Uses Other Columns?** | ❌ No | ✅ Yes |
| **Captures Relationships** | ❌ Ignores feature correlations | ✅ Preserves relationships between variables |
| **Accuracy** | Lower when features are correlated | Higher when features are correlated |
| **Speed** | Fast | Slower (iterative process) |
| **Complexity** | Simple | More computationally expensive |

### Univariate Imputation
- Treats each column separately.
- Common strategies:
  - Mean
  - Median
  - Mode (Most Frequent)
  - Constant value
- Best when missing values are random and features are largely independent.

### Multivariate Imputation
- Predicts missing values using other features.
- Learns relationships between variables.
- Common methods:
  - `IterativeImputer` (MICE)
  - KNN Imputer
- Best when features are correlated and preserving data structure is important.

### Rule of Thumb
- **Use Univariate Imputation** → Simple datasets, quick preprocessing, low correlation between features.
- **Use Multivariate Imputation** → When features influence each other and you want more accurate imputations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('titanic_toy.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().mean()*100
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train.shape, X_test.shape
```

```python
X_train.isnull().mean()
```

```python
mean_age = X_train['Age'].mean()
median_age = X_train['Age'].median()

mean_fare = X_train['Fare'].mean()
median_fare = X_train['Fare'].median()
```

```python
X_train['Age_median'] = X_train['Age'].fillna(median_age)
X_train['Age_mean'] = X_train['Age'].fillna(mean_age)

X_train['Fare_median'] = X_train['Fare'].fillna(median_fare)
X_train['Fare_mean'] = X_train['Fare'].fillna(mean_fare)
```

```python
X_train.sample(5)
```

```python
print('Original Age variable variance: ', X_train['Age'].var())
print('Age Variance after median imputation: ', X_train['Age_median'].var())
print('Age Variance after mean imputation: ', X_train['Age_mean'].var())

print('Original Fare variable variance: ', X_train['Fare'].var())
print('Fare Variance after median imputation: ', X_train['Fare_median'].var())
print('Fare Variance after mean imputation: ', X_train['Fare_mean'].var())
```

variance will shrink after mean, median imputation

```python
fig = plt.figure()
ax = fig.add_subplot(111)

# original variable distribution
X_train['Age'].plot(kind='kde', ax=ax)

# variable imputed with the median
X_train['Age_median'].plot(kind='kde', ax=ax, color='red')

# variable imputed with the mean
X_train['Age_mean'].plot(kind='kde', ax=ax, color='green')

# add legends
lines, labels = ax.get_legend_handles_labels()
ax.legend(lines, labels, loc='best')
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

# original variable distribution
X_train['Fare'].plot(kind='kde', ax=ax)

# variable imputed with the median
X_train['Fare_median'].plot(kind='kde', ax=ax, color='red')

# variable imputed with the mean
X_train['Fare_mean'].plot(kind='kde', ax=ax, color='green')

# add legends
lines, labels = ax.get_legend_handles_labels()
ax.legend(lines, labels, loc='best')
```

### Using Sklearn

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
imputer1 = SimpleImputer(strategy='median')
imputer2 = SimpleImputer(strategy='mean')
```

```python
trf = ColumnTransformer([
    ('imputer1',imputer1,['Age']),
    ('imputer2',imputer2,['Fare'])
],remainder='passthrough')
```

```python
trf.fit(X_train)
```

```python
trf.named_transformers_['imputer1'].statistics_
```

```python
trf.named_transformers_['imputer2'].statistics_
```

```python
X_train = trf.transform(X_train)
X_test = trf.transform(X_test)
```
