# Multivariate Imputation by Chained Equations (MICE)

* Used only on Data Missing At Random

```python
import numpy as np
import pandas as pd
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
df = np.round(pd.read_csv('Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)
np.random.seed(9)
df = df.sample(5)
df
```

```python
df = df.iloc[:,0:-1]
df
```

```python
df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan
df
```

```python
imputer = IterativeImputer(estimator=LinearRegression(), imputation_order='roman', random_state=0, verbose=2)
```

```python
imputed_data = imputer.fit_transform(df)
```

```python
df_imputed = np.round(pd.DataFrame(imputed_data, columns=df.columns, index=df.index))
```
