# 3. Opening a csv file from an URL

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

# 10. Encoding parameter

```python
# Detect which encoding
from charset_normalizer import from_bytes

with open("zomato.csv", "rb") as f:
    raw_data = f.read()
    result = from_bytes(raw_data)
    print(result.best().encoding)
```

```python
pd.read_csv('zomato.csv',encoding='latin')
```

# 11. Skip bad lines

```python
# error_bad_lines is removed in newer pandas versions,
# error_bad_lines is replaced by on_bad_lines

# pd.read_csv('BX-Books.csv', sep=';', encoding="latin-1", error_bad_lines=False)

pd.read_csv('BX-Books.csv', sep=';', encoding="latin-1", on_bad_lines = 'skip' )
```

# 13. Handling Dates

```python
pd.read_csv('IPL Matches 2008-2020.csv',parse_dates=['date']).info()
```

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

# 14. Convertors

```python
pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename})
```
